# Chấm sentiment bằng transformer

Notebook này chỉ là hướng dẫn chạy; phần tính toán nằm trong `src/sentiment_transformer.py`.

## Cách chạy

Chạy từ thư mục gốc của dự án, trong terminal (không phải trong notebook):

```
pip install torch transformers
python -m src.sentiment_transformer score
python -m src.sentiment_transformer report
```

- `score`: chấm điểm theo từng khối 8192 review, lưu checkpoint vào `data/processed/<CATEGORY>/tf_cache_len<max-len>/`. Chạy lại sau khi bị ngắt sẽ tiếp tục từ khối đang dở. Nếu văn bản đầu vào thay đổi (ví dụ bạn chạy lại `prepare_and_eda`), cache cũ tự bị xóa.
- `report`: in chẩn đoán điểm so với số sao (toàn bộ review và riêng review từ 15 từ trở lên) và ghép cột `tf01` vào `sentiment_all.csv.gz`.
- Chỉ có CPU thì thêm `--max-len 128` để nhanh hơn. Thời gian phụ thuộc số review (`MAX_REVIEWS` ở bước chuẩn bị dữ liệu).
